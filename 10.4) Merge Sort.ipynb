#### Merge Sort
Merge Sort is one of the most important sorting algorithms in DSA. It follows the Divide and Conquer strategy.

The main idea is:
1. Divide the problem into smaller parts
2. Solve the smaller parts
3. Combine (merge) the solutions

Merge Sort was created, because Algorithms like: Bubble Sort, Selection Sort, Insertion Sort have average complexity: O(n²).

So, as the amount of data grows, they become slow. Merge sort was designed to achieve O(n log n) which is much faster for large datasets.

```python
# Code for merge sort. Here we are creating new array all the time and then return. So, it is not optimal in terms of space complexity.
def merge_two_sorted_lists(a, b):
    sorted_list = []
    len_a = len(a)
    len_b = len(b)
    # Pointers for both lists
    i = j = 0
    # Compare elements from both lists and pick the smaller one
    while i < len_a and j < len_b:
        if a[i] <= b[j]:
            sorted_list.append(a[i])
            i += 1
        else:
            sorted_list.append(b[j])
            j += 1

    # Add remaining elements from list a (if any)
    while i < len_a:
        sorted_list.append(a[i])
        i += 1
    # Add remaining elements from list b (if any)
    while j < len_b:
        sorted_list.append(b[j])
        j += 1
    return sorted_list


def merge_sort(arr):
    # Base Case: Array with 0 or 1 element is already sorted
    if len(arr) <= 1:
        return arr
    # Divide array into two halves
    mid = len(arr) // 2
    left = arr[:mid]
    right = arr[mid:]
    # Recursively sort both halves
    left = merge_sort(left)
    right = merge_sort(right)
    # Merge the two sorted halves
    return merge_two_sorted_lists(left, right)

if __name__ == '__main__':
    # a = [5,8,12,56]
    # b = [7,9,11,15]
    # print(merge_two_sorted_lists(a,b))
    arr = [10,3,15,7,8,23,98,29]
    print(merge_sort(arr))
```

```python
# Similar Code for merge sort. But we will only modify the original list directly here rather than returning the sorted lists back for merging.
def merge_two_sorted_lists(a, b, arr):
    # Length of left and right sub-arrays
    len_a = len(a)
    len_b = len(b)
    # i -> left array pointer
    # j -> right array pointer
    # k -> original array pointer
    i = j = k = 0
    # Compare elements from both halves and place smaller one into arr
    while i < len_a and j < len_b:
        if a[i] <= b[j]:
            arr[k] = a[i]
            i += 1
        else:
            arr[k] = b[j]
            j += 1
        k += 1
    # Copy remaining elements from left half
    while i < len_a:
        arr[k] = a[i]
        i += 1
        k += 1
    # Copy remaining elements from right half
    while j < len_b:
        arr[k] = b[j]
        j += 1
        k += 1


def merge_sort(arr):
    # Base Case: Already sorted
    if len(arr) <= 1:
        return arr
    # Divide array into two halves
    mid = len(arr) // 2
    left = arr[:mid]
    right = arr[mid:]
    # Recursively sort both halves
    merge_sort(left)
    merge_sort(right)
    # Merge sorted halves back into original array
    merge_two_sorted_lists(left, right, arr)
    return arr

if __name__ == '__main__':
    # a = [5,8,12,56]
    # b = [7,9,11,15]
    # print(merge_two_sorted_lists(a,b))
    arr = [10,3,15,7,8,23,98,29]
    merge_sort(arr)
    print(arr)

# Note: First Version,  Extra memory from:left,right,sorted_list
# Overall: O(n)
# Second Version: Extra memory from: left, right , but No sorted_list.
# Overall: O(n).  Still same for second:O(n)
# The second version only uses slightly less memory because it doesn't create:sorted_list = []
```
