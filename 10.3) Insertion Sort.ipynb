#### Insertion Sort
Insertion Sort is one of the simplest sorting algorithms. The idea comes from how humans sort playing cards in their hands.

- Real-Life Example

Suppose you pick cards one by one:

5  2  4  6  1

You don't throw all cards randomly and sort them later.

Instead:

- Take first card → already sorted.
- Take second card → insert at correct position.
- Take third card → insert at correct position.
- Continue.

This is exactly how Insertion Sort works.

Left side = Sorted   &  Right side = unsorted

- We consider 5 as sorted part, and then the algorithm picks the first element from the unsorted part(remaining array on left side) and inserts it into its correct position in the sorted part.
- [5 | 2 4 6 1].  --   Compare 2<5
- So, move 5 right and insert 2.     -- [2 5 | 4 6 1]
- Compare 4<5. So, again sift 5 and insert 4 after 2.    --    [2 4 5 | 6 1]
- Compare 6>5. So, 6 already in correct position and no shifting required.    --   [2 4 5 6 | 1]
- Compare 1<6,1<5,1<4,1<2. Shift everything right and insert 1. Now, it is sorted.   --    [1 2 4 5 6]

##### Time complexity
- Worst case - O(n²) comparisons and swaps
- Best Case - O(n) comparison, O(1) swaps
- Average - O(n²) comparisons and swaps
- Worst case space complexity - O(n) total, O(1) auxillary)

```python
# Creating program for insertion sort
def insertion_sort(elements):

    # Start from index 1 because a single element at index 0 is already considered sorted.
    for i in range(1, len(elements)):
        # Store the current element that needs to be inserted into its correct position in the sorted portion.
        anchor = elements[i]

        # Start comparing with the element immediately to its left
        j = i - 1

        # Keep moving left while: 1. We haven't gone beyond index 0 # 2. The left element is greater than anchor
        while j >= 0 and elements[j] > anchor:
            # Shift the larger element one position to the right
            elements[j + 1] = elements[j]
            # Move further left to continue checking
            j = j - 1

        # Insert anchor into its correct position j+1 is used because j has already moved one step too far when the while loop stops.
        elements[j + 1] = anchor
        pass #  pass is a placeholder statement. It has no effect here and can be safely removed.

if __name__ == '__main__':

    tests = [
        [11,9,29,7,2,15,28],
        [3, 7, 9, 11],
        [25, 22, 21, 10],
        [29, 15, 28],
        [],
        [6]
    ]

    for elements in tests:
        insertion_sort(elements)
        print(f'sorted array: {elements}')

```
