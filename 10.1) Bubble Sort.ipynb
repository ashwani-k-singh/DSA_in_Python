#### 10.1) Bubble Sort:
Bubble Sort repeatedly compares adjacent elements and swaps them if they are in the wrong order. The largest element "bubbles up" to the end after each pass.

###### Characteristics
- Simple to understand
- Many swaps
- Poor performance on large datasets

In the above image we can see the highest value(in short bubble) will pop up and reach the end.  Means the bubble in the water went to the topmost position.
- Process 2 (Repeat same again)

So we will continue the above proces until the list is sorted. For this 2 for loop will run together.

So, Time Complexity - O(n)

Space Complexity - O(1)

Space complexity is O(1), since we are using the same space and just swapping the elements. Only one additional variable is used for swapping.

```python
# Bubble sort in python

def bubble_sort(elements):
    size = len(elements)
    for i in range(size-1):          #run from 0 to (size-2). We need to run till the last 2nd element, it will auto compare with last element
        swapped = False
        for j in range(size-1-i):    # we are using size-1-i, because when in ist loop max element comes at top, so no need to do iteration on it in next loop
            if elements[j] > elements[j+1]:
                ref_var = elements[j+1]
                elements[j+1] = elements[j]
                elements[j] = ref_var
                swapped = True
        if swapped == False:      #if no swap in my first full loop run for i=0, it means my data is already sorted.
            break
    # we didn't pass return here,since both elements and element_list both refer to the same list object and lists are mutable in Python.


if __name__ == '__main__':
    element_list = [5,9,2,1,67,34,88,34]
    bubble_sort(element_list)
    print(element_list)
```

```python
def bubble_sort(elements,key):
    size = len(elements)
    for i in range(size-1):
        swapped = False
        for j in range(size-1-i):
            if elements[j][key] > elements[j+1][key]:
                ref_var = elements[j+1]
                elements[j+1] = elements[j]
                elements[j] = ref_var
                swapped = True
        if not swapped:  #same as swapped == False
            break

if __name__ == '__main__':
    elements = [
        { 'name': 'mona',   'transaction_amount': 1000, 'device': 'iphone-10'},
        { 'name': 'dhaval', 'transaction_amount': 400,  'device': 'google pixel'},
        { 'name': 'kathy',  'transaction_amount': 200,  'device': 'vivo'},
        { 'name': 'aamir',  'transaction_amount': 800,  'device': 'iphone-8'},
    ]
    print(elements[0]['name']) #return the name at 0 index from dict.
    bubble_sort(elements,'name')
    print(elements)
```

```python
bubble_sort(elements,'transaction_amount')
print(elements)
```
