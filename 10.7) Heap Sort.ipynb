##### Heap Sort:
Heap Sort is a sorting algorithm that uses a special tree structure called a Heap. A Heap is a Complete Binary Tree that follows a special rule.
1. Max Heap - Parent is always greater than or equal to its children. In short, Largest element is always at the root.


2. Min Heap - Parent is always smaller than or equal to its children. Smallest element is at the root.
