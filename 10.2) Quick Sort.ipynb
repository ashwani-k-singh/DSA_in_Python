First, we will create a Quick sort program, then we will learn the hoare parition and lomuto partition.
#### Quick Sort code. Here we will be creating new lists to store the results.
Steps to Solve:
1. Pick pivot (first element of the current array/sub-array).
2. Create two empty lists.
    - smaller - stores elements less than the pivot.
    - greater - stores elements greater than or equal to the pivot.
3. Traverse the remaining elements (skip the pivot).
    - We start from index 1 because index 0 is already selected as the pivot.
4. Compare each element with the pivot.
    - If element < pivot, add it to smaller.
    - Otherwise, add it to greater.
5. Partition is complete.
    - smaller contains all elements less than the pivot.
    - greater contains all elements greater than or equal to the pivot.
6. Recursively sort the smaller list.
7. Recursively sort the greater list.
8. Combine the results.
    - Place the pivot between the two sorted lists.
9. Repeat recursively for every new sub-array created.
10. Stop recursion when: len(arr) <= 1 , because an array with 0 or 1 element is already sorted.

<img src=".\Reference Images\qick sort.png" width="600">


```python
# Quick Sort
# Creates new lists instead of sorting in-place.
# Best for understanding recursion and partitioning.

def quick_sort(arr):
    # Base Case:
    # If array has 0 or 1 element, it is already sorted.
    if len(arr) <= 1:
        return arr

    # Choose first element as pivot
    pivot = arr[0]

    # Elements smaller than pivot
    smaller = []

    # Elements greater than pivot
    greater = []

    # Traverse remaining elements
    for element in arr[1:]:

        # Put smaller values into left list
        if element < pivot:
            smaller.append(element)

        # Put greater or equal values into right list
        else:
            greater.append(element)

    # Recursively sort left and right parts
    #
    # quick_sort(smaller)
    # +
    # [pivot]
    # +
    # quick_sort(greater)
    #
    return quick_sort(smaller) + [pivot] + quick_sort(greater)


if __name__ == '__main__':

    arr = [9,3,7,1,8,2,5]
    sorted_arr = quick_sort(arr)
    print("Original :", arr)
    print("Sorted   :", sorted_arr)
```

#### Hoare Partition:
The goal is to place smaller elements on the left side of the pivot and larger elements on the right side of the pivot.

Unlike Lomuto Partition, the pivot is not guaranteed to be in its final sorted position after partitioning.
##### Steps to solve (General steps)
1. Choose a pivot (usually the first element).
2. Create two pointers:
    - i starts from the left side.
    - j starts from the right side.
3. Move i to the right until an element >= pivot is found.
4. Move j to the left until an element <= pivot is found.
5. If i < j, swap the elements at i and j because they are on the wrong sides of the pivot.
6. Repeat Steps 3 to 5 until the pointers meet or cross (i >= j).
7. When i >= j, partitioning is complete.
8. Recursively apply the same process to the left and right partitions.



Recursive tree for the next code which we will write using Hoare method:

Steps for our code:
1. Pick the pivot (first element of the current sub-array).
2. Move the start pointer to the right until a value greater than the pivot is found.
    - start initially begins at the pivot index.
    - Since pivot <= pivot is always true, the first comparison immediately advances start to the next element.
    - This prevents the pointer from stopping at the pivot itself.
3. Move the end pointer to the left until a value less than or equal to the pivot is found.
4. If start < end, swap the elements at the start and end pointers because they are on the wrong sides of the pivot.
5. Repeat Steps 2–4 while start < end. The loop ends when the pointers meet or cross (start >= end).
6. Swap the pivot with the element at the end pointer. This places the pivot in its correct sorted position.
7. Partition is complete.
    - The pivot is now in its final sorted position.
    - All elements to the left of the pivot are <= pivot.
    - All elements to the right of the pivot are > pivot.
8. Recursively sort the left partition.
9. Recursively sort the right partition.
10. Stop recursion when start >= end, because the current sub-array contains 0 or 1 element and is already sorted.


[9, 3, 7, 1, 8, 2, 5]


```python
# Quick Sort using a two-pointer partition(hoare) approach
# Pivot = First element of the current sub-array
# Here, in this code we are not creating new lists, we are only rearranging the same array in place.

def swap(a, b, arr):
    # Swap two elements in the array using their indices
    if a != b:   # No need to swap if both indices are same
        tmp = arr[a]
        arr[a] = arr[b]
        arr[b] = tmp

def quick_sort(elements, start, end):
    """
    Quick Sort recursively sorts the left and right portions
    around a pivot element.
    start = starting index of current sub-array
    end   = ending index of current sub-array
    """
    # Base condition: If start >= end, sub-array has 0 or 1 element and is already sorted.
    if start < end:
        # Partition the array and get pivot's final position
        pi = partition(elements, start, end)

        # Sort all elements left of pivot. recursively call until null
        quick_sort(elements, start, pi - 1)

        # Sort all elements right of pivot
        quick_sort(elements, pi + 1, end)


def partition(elements, start, end):
    # Select first element as pivot
    pivot_index = start
    pivot = elements[pivot_index]

    # Move start and end pointers toward each other
    while start < end:
        # Move start pointer right until finding a value greater than pivot
        while start < len(elements) and elements[start] <= pivot:
            start += 1

        # Move end pointer left until finding a value smaller than or equal to pivot
        while elements[end] > pivot:
            end -= 1

        # If pointers haven't crossed,swap the misplaced elements
        if start < end:
            swap(start, end, elements)

    # At this point:end points to the correct location for pivot
    # Example: Before: [9,3,7,1,8,2,5]
    # After pointer movement: [5,3,7,1,8,2,9]
    # Put pivot (9) in its final position
    swap(pivot_index, end, elements)

    # Return pivot's final index
    return end


if __name__ == '__main__':

    elements = [9, 3, 7, 1, 8, 2, 5]

    quick_sort(elements, 0, len(elements) - 1)

    print(elements)

```

#### Lomuto Patition
The goal is to place the pivot in its correct sorted position and ensure:

   - All elements smaller than or equal to pivot are on the left side.
   - All elements greater than pivot are on the right side.

Unlike Hoare Partition, the pivot is guaranteed to be in its final sorted position after partitioning.


<img src=".\Reference Images\lomuto.png" width="500">


Proper steps for the below code:
1. Choose a pivot (last element of the current sub-array).
    - pivot = elements[end]
2. Create a partition index (p_index).
    - p_index represents the position where the next element smaller than or equal to the pivot should be placed.
    - p_index = start
3. Traverse the array from left to right (excluding the pivot).
    - for i in range(start, end):
4. Compare each element with the pivot.
    - elements[i] <= pivot
5. If the current element is smaller than or equal to the pivot:
    - Swap elements[i] with elements[p_index]. swap(i, p_index, elements)
    - Increment p_index. p_index += 1
6. If the current element is greater than the pivot:
    - Do nothing.
    - Continue scanning the next element.
7. Continue until all elements before the pivot have been checked.
8. After traversal is complete, swap the pivot with elements[p_index].
    - swap(p_index, end, elements)
9. The pivot is now in its correct sorted position.
    - All elements left of pivot <= pivot
    - All elements right of pivot > pivot
10. Return the pivot index.
    - return p_index
11. Recursively apply the same process to the left and right partitions.
    - quick_sort(elements, start, p_index-1)
    - quick_sort(elements, p_index+1, end)

```python
# Lomuto partitioning code
# Quick Sort using Lomuto Partition Scheme
# Pivot = Last element of the current sub-array

def swap(a, b, arr):
    # Swap elements at index a and b No swap needed if both indices are same
    if a != b:
        tmp = arr[a]
        arr[a] = arr[b]
        arr[b] = tmp


def quick_sort(elements, start, end):

    # Base Case: If start >= end, sub-array has 0 or 1 element and is already sorted.
    if start < end:
        # Partition the array and get pivot's final position
        pi = partition(elements, start, end)
        # Recursively sort left side of pivot
        quick_sort(elements, start, pi - 1)
        # Recursively sort right side of pivot
        quick_sort(elements, pi + 1, end)


def partition(elements, start, end):
    # Lomuto chooses the LAST element as pivot
    pivot = elements[end]
    # Partition Index: Keeps track of where the next element <= pivot should be placed.
    p_index = start

    # Traverse all elements except the pivot
    for i in range(start, end):
        # If current element belongs on left side
        if elements[i] <= pivot:
            # Move element to correct position
            swap(i, p_index, elements)
            # Expand left partition boundary
            p_index += 1
    # Place pivot between left and right partitions
    swap(p_index, end, elements)
    # Return pivot's final position
    return p_index

if __name__ == '__main__':
    tests = [
        [11,9,29,7,2,15,28],
        [3, 7, 9, 11],
        [25, 22, 21, 10],
        [29, 15, 28],
        [],
        [6]
    ]
    # elements = ["mona", "dhaval", "aamir", "tina", "chang"]

    for elements in tests:
        quick_sort(elements, 0, len(elements)-1)
        print(f'sorted array: {elements}')
```

Note: pi stands for Partition Index. It is the index returned by the partition() function and tells Quick Sort: The final position of the pivot after partitioning.
