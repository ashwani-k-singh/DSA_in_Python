## Sorting:
Sorting means arranging data in a particular order. Whether Ascending or descending.

We sort the data to make search faster. E.g. Binary search technique.
###### Different Sorting Techniques:
- Bubble Sort (Bubble Sort repeatedly compares adjacent elements and swaps them if they are in the wrong order. The largest element "bubbles up" to the end after each pass.)
- Quick sort (Choose a Pivot and partition elements around it.)
- Insertion sort (Build the sorted portion one element at a time. Similar to arranging cards in your hand.)
- Merge sort (Uses the Divide and Conquer technique.)
- Shell Sort (Improved version of Insertion Sort. Instead of comparing adjacent elements only, it compares elements separated by a gap.)
- Selection sort (Find the smallest element and place it at the beginning. Then find the second smallest and place it at the second position. Continue until sorted.)
- Heap sort (Uses a special tree structure called a Heap. A Max Heap always keeps the largest element at the root.)
